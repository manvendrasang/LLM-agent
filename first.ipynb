```python
with open('wizard.txt', 'r', encoding='utf-8') as f:
    text = f.read()
# print(len(text))
# print(text[:200])
chars = sorted(set(text))
# print(chars)
# print(len(chars))
```

```python
string_to_int = { ch:i for i, ch in enumerate(chars) }
int_to_string = { i:ch for i, ch in enumerate(chars) }
encode = lambda s: [string_to_int[c] for c in s]
decode = lambda l: ''.join([int_to_string[i] for i in l])
# print(encode('hello'))
# encoded = encode('hello')
# decoded = decode(encoded)
# print(decoded)
data = torch.tensor(encode(text), dtype=torch.long)
print(data[:100])
```

```python
n = int(0.8*len(data))
train_data = data[:n]
val_data = data[n:]
```

```python
block_size = 8
x = train_data[:block_size]
y = train_data[1:block_size+1]

for t in range(block_size):
    context = x[:t+1]
    target = y[t]
    print(f"when input is {context.tolist()} the target: {target}")
```

```python
torch_rand1 = torch.rand(10000, 10000, device=device)
torch_rand2 = torch.rand(10000, 10000, device=device)
np_rand1 = torch.rand(10000, 10000)
np_rand2 = torch.rand(10000, 10000)

start_time = time.time()

rand = (torch_rand1 @ torch_rand2)

end_time = time.time()

elapsed_time = end_time - start_time
print(f"torch rand: {elapsed_time: 10f}")

start_time = time.time()

rand = np.multiply(np_rand1, np_rand2)

end_time = time.time()
elapsed_time = end_time - start_time
print(f"numpy rand: {elapsed_time: 10f}")
```
